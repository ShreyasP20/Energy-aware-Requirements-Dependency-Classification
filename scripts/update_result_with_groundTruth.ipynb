```python
import pandas as pd

file1 = ""
file2 = ""

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)


df1.rename(columns={"text1": "Text1", "text2": "Text2"}, inplace=True)
df2 = df2.loc[:, ~df2.columns.str.contains("^Unnamed")]

direct_map = {
    (row.Text1, row.Text2): row.Class
    for _, row in df2.iterrows()
}

reverse_map = {
    (row.Text2, row.Text1): row.Class
    for _, row in df2.iterrows()
}

def resolve_ground_truth(row):
    key = (row.Text1, row.Text2)

    if key in direct_map:
        return direct_map[key]

    if key in reverse_map:
        return reverse_map[key]

    return row.ground_truth

df1["ground_truth"] = df1.apply(resolve_ground_truth, axis=1)

output_file = "updated_first_csv_bidirectional.csv"
df1.to_csv(
    "updated_first_csv_bidirectional.csv",
    index=False,
    float_format="%.12f"
)

print("Ground truth updated using direct + bidirectional matching.")
```
